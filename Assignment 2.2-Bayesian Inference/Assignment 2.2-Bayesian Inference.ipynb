## Conditional Probability Model and Classification

In machine learning, we're often interested in a predictive modeling problem where we want to predict a class label for a given observation. For example, we might want to classify the species of plants based on the measurements of the flowers.

Problems of this type are referred to as classification predictive modeling problems. The observation or input to the model is referred to as `X`, and the class label or output of the model is referred to as `y`.

Together, `X` and `y` represent observations collected from the domain, i.e., a table or matrix (columns and rows or features and samples) of training data used to fit a model. The model must learn how to map specific examples to class labels or $y = f(X)$, which minimizes the misclassification.


One approach to solving this problem is to develop a **probabilistic model**. From a probabilistic perspective, we are interested in estimating the conditional probability of the class label, given the observation.

For example, a classification problem may have $k$ class labels $y_1, y_2, \dots, y_k$ and $n$ input variables, $X_1, X_2, \dots, X_n$. We can calculate the conditional probability for a class label with a given instance or set of input values for each column $X_1, X_2, \dots, X_n$ as:

$$P(y_i | X_1, X_2, \dots, X_n).$$


Bayes' theorem provides a principled way of calculating this conditional probability.

A simple form of the calculation for Bayes' theorem is as follows:

$$P(A|B) = P(B|A) * P(A) / P(B),$$

where the probability that we are interested in calculating, $P(A|B),$ is called the posterior probability, and the marginal probability of the event $P(A)$ is called the prior.

## Dataset Introduction

The dataset contains the following columns, each representing a specific characteristic of different wine samples:

1. **Class**: Categorical identifier indicating the class of wine. This is likely a label used for classification tasks.

2. **Alcohol**: Represents the alcohol content in the wine, measured as a percentage by volume. It is an important feature affecting the taste and quality of wine.

3. **Malic Acid**: Indicates the concentration of malic acid in the wine, which influences the tartness and is a key component in the wine's overall flavor profile.

4. **Ash**: Reflects the total ash content in the wine, which can affect color and taste. Ash content is a measure of the mineral content of the wine.

5. **Alcalinity of Ash**: Measures the alkalinity (as opposed to acidity) of the ash found in the wine, which can influence the stability and color of the wine.

6. **Magnesium**: The level of magnesium in the wine. Magnesium can influence the growth of vines and the health of the wine.

7. **Total Phenols**: Represents the total amount of phenols in the wine. Phenols are compounds that affect the taste, color, and mouthfeel of wine.

8. **Flavanoids**: A type of phenol specifically known for its antioxidant properties, which contribute to the body and structure of the wine as well as its longevity.

9. **Nonflavanoid Phenols**: These are phenols that do not include flavanoids, but still contribute to the sensory characteristics of wine.

10. **Proanthocyanins**: These are tannins that provide flavor, color, and protection against oxidation, contributing significantly to the wine's taste and longevity.

11. **Color Intensity**: This is a measurement of the wine's color depth, which can give hints about the age and concentration of the grapes used.

12. **Hue**: Indicates the type of color tint or shade visible in the wine, which can provide insights into the age and grape variety.

13. **OD280/OD315 of Diluted Wines**: This is a ratio used to measure the wine's absorbance at 280 nm and 315 nm, indicative of protein stability and overall wine quality.

14. **Proline**: Represents the amount of proline, an amino acid, present in the wine, which can be a marker of grape maturity and wine quality.

These columns collectively provide a comprehensive chemical profile of wine samples, which can be used for tasks such as classification, quality assessment, and understanding the relationships between different chemical properties and wine characteristics.

## Implement Gaussian Naive Bayes

Gaussian Naive Bayes is a classification algorithm used in machine learning. It works on the principle of Bayes' Theorem, assuming that features are independent of each other. Gaussian Naive Bayes specifically deals with continuous data and assumes that the data follows a normal (Gaussian) distribution.

The Gaussian Naive Bayes works as follows:  
1.Computation of Prior Probabilities: Determine the proportion of each class in the training data.

2. Computation of Mean and Variance: For each feature and class, calculate the mean and variance.

3. Computation of Likelihood: For a new data point, compute the likelihood of each feature value based on the Gaussian distribution.

4. Calculate Posterior Probability: Combine these likelihoods with the prior probabilities to get the posterior probability for each class.

5. Predict Class: The class with the highest posterior probability is the predicted class.

## Implementation of Gaussian Naive Bayes in scikit-learn

The GaussianNB() class is imported from sklearn.naive_bayes, and is used to train the model.

```python
#Initialize and train the Gaussian Naive Bayes model.
from sklearn.naive_bayes import GaussianNB
nbc = GaussianNB()
nbc.fit(X_train,y_train)
```

```python
#perform prediction on test data
y_pred = nbc.predict(X_test)
```

Once the model is trained, the model can be used to predict the test dataset. But the acceptance of the model, depends on how good it is able to predict the test samples accurately.

## Model Evaulation

 The accuracy of the predicted model is computed using the accuracy_score metric of `sklearn`

```python
from sklearn.metrics import accuracy_score

# y_true are the true class labels
# y_pred are the predicted class labels by the classifier
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```
